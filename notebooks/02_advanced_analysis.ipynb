## Compliance check and report generation (Markdown + HTML)

```python
comp = checker.check(clauses, 'service')
print('Compliance score:', comp['compliance_score'])
md = reporter.generate_markdown({
    'filename': 'sample_contract.txt',
    'text_length': len(sample_text),
    'clauses_found': len(clauses),
    'clauses': clauses,
    'risk_summary': risk_result,
    'compliance': comp,
})
print('
--- MARKDOWN PREVIEW ---
')
print(md[:1000])
# Optionally write HTML to a file
html_path = reporter.generate_html({
    'filename': 'sample_contract.txt',
    'clauses': clauses,
    'risk_summary': risk_result,
    'compliance': comp,
}, output_path='advanced_report.html')
print('
HTML report saved to', html_path)
```
